```python
import pandas as pd

from scimmia.store import load_draws_df, load_pools_df, load_prizes_df
from scimmia.stats import descriptive as ds, honesty as hs

draws = load_draws_df()
prizes = load_prizes_df()
pools = load_pools_df()
print(len(draws), 'estrazioni dal', draws.date.min().date(), 'al', draws.date.max().date())
draws.tail()
```

## Frequenze e ritardi

```python
freq = ds.frequency(draws)
freq.sort_values(ascending=False).head(10)
```

```python
ds.delays(draws).sort_values('current', ascending=False).head(10)
```
